# Identification des Sources de Données

## A. Base de données ben_telecom
Le système source principal est la base de donnée `ben_telecom`.

Un accès  en mode lecture nous a été accordée sur une copie indépendante de cette base de donnée `ben_telecom_dev`, destinée à la phqse de développement.
### 1. Se connecter à la base de données
```sh
psql -d ben_telecom_dev;
```

### 2. Lister les schémas
```sh
\dn+
```

On a les shémas suivants:
- rm
- subscriptions
- offers_catalog


### 3. Lister les tables de chaque schéma

#### 3.1. Pour le shéma offers_catalog
```sh
\dt+ offers_catalog.*
```

On a les tables
- offers
- offer_billings
- et offer_configurations

### 3.2. Pour le schéma crm

```sh
\dt+ crm.*
```

On a la table customers

### 3.3 Pour le schéma subscriptions

```sh
\dt+ subscriptions.*
```

On a la table subscriptions

## C. Système de Médiation d'Usage (Kafka)

### 0. Démarer le serveur Kafka (S'il n'est pas actif auparavant)

```sh
cd data
```

```sh
./kafka/bin/kafka-server-start.sh ./kafka/config/server.properties
```

### 1. Lister les topics
```sh
./kafka/bin/kafka-topics.sh  --bootstrap-server localhost:9092 --list
```

TelecomUsage est le seul topic

#### 2.  Décrie le topic Telecom Usage

```sh
./kafka/bin/kafka-topics.sh --describe --topic TelecomUsage --bootstrap-server localhost:9092
```
